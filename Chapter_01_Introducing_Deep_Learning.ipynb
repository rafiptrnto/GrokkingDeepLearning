## 1. Why you should learn deep learning

Buku menyebut deep learning sebagai *alat untuk otomatisasi kecerdasan
secara bertahap*. Menurut penulis, deep learning adalah lanjutan dari
usaha manusia membuat alat yang makin baik untuk memahami dan
mengendalikan lingkungannya.

Beberapa poin dari bagian ini:

- **Pencapaian.** Deep learning mencatat hasil baru pada computer vision,
  pengenalan suara, dan terjemahan mesin. Metode yang dipakai hampir sama
  (neural network yang terinspirasi otak) untuk banyak bidang tersebut.
- **Otomatisasi pekerjaan.** Penulis menyebut kemungkinan *job displacement*:
  bahkan jika kemajuan berhenti hari ini, dampaknya pada pekerjaan terampil
  sudah besar. Contoh yang disebut: operator call-center, sopir taksi, dan
  analis bisnis tingkat dasar.
- **Menyenangkan dan kreatif.** Mensimulasikan kecerdasan dan kreativitas
  mengajarkan tentang manusia itu sendiri. Contoh buku: melatih mesin
  melukis seperti Monet dari dataset semua lukisan.
