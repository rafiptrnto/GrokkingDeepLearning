## 1. What is deep learning?

Deep learning adalah **subset dari metode machine learning**. Machine
learning sendiri adalah bidang yang mempelajari dan mengembangkan mesin
yang dapat belajar. Deep learning terutama memakai *artificial neural
network*, kelas algoritma yang terinspirasi secara longgar oleh otak manusia.

Di industri, deep learning dipakai untuk computer vision (gambar),
natural language processing (teks), dan automatic speech recognition (audio).
Buku ini berfokus pada dasar-dasar deep learning yang dipakai baik di riset
maupun di industri.

## 2. What is machine learning?

Definisi yang dipakai buku: *a field of study that gives computers the
ability to learn without being explicitly programmed*. Mesin mengamati
sebuah pola lalu mencoba menirunya, secara langsung atau tidak langsung.
Buku menggambarkan ini sebagai *monkey see, monkey do*.

Ada dua jenis utama: **supervised** (peniruan langsung) dan
**unsupervised** (peniruan tidak langsung).

## 3. Supervised machine learning

Supervised learning adalah metode untuk **mengubah satu dataset menjadi
dataset lain**. Contoh pada buku: dataset *Monday Stock Prices* (harga saham
setiap Senin selama 10 tahun) dan *Tuesday Stock Prices*. Algoritma dapat
dilatih memakai salah satunya untuk memprediksi yang lain.

Contoh tugas supervised yang disebut buku:
- Dari piksel gambar, mendeteksi ada atau tidaknya kucing
- Dari film yang disukai, memprediksi film lain yang mungkin disukai
- Dari kata-kata seseorang, memprediksi apakah ia senang atau sedih
- Dari berita, memprediksi harga saham besok
- Dari audio, memprediksi transkripnya

Buku menyebut supervised learning sebagai "roti dan mentega" AI terapan
(narrow AI). Sebagian besar pekerjaan machine learning menghasilkan
*supervised classifier*, dan unsupervised learning biasanya dipakai untuk
membantu membangun classifier supervised yang akurat.

## 4. Unsupervised machine learning

Unsupervised learning juga mengubah satu dataset menjadi dataset lain, tetapi
dataset hasilnya **tidak diketahui sebelumnya** dan tidak ada "jawaban benar"
yang ditiru. Algoritma diminta menemukan pola dan menceritakannya.

Contoh utama adalah **clustering**: daftar titik data diubah menjadi daftar
label cluster (misalnya 1–10). Algoritma tidak memberi tahu nama tiap cluster.
Buku menyatakan bahwa semua bentuk unsupervised learning dapat dipandang
sebagai bentuk clustering.

## 5. Parametric vs. nonparametric learning

Buku membagi algoritma dengan cara kedua, yaitu parametrik atau nonparametrik.
Kedua pembagian ini saling bebas, sehingga ada empat kombinasi: supervised
parametrik, unsupervised parametrik, supervised nonparametrik, dan
unsupervised nonparametrik.

Definisi dari buku:
- **Parametric model**: jumlah parameternya **tetap**.
- **Nonparametric model**: jumlah parameternya **tak terbatas** (ditentukan oleh data).

Analogi buku: menyambungkan kotak persegi ke lubangnya. Bayi memasukkannya ke
semua lubang sampai pas (parametrik, trial and error). Remaja menghitung sisinya
lalu mencari lubang dengan jumlah sisi yang sama (nonparametrik, menghitung).

Buku juga mencatat ada perdebatan tentang batasan pasti kedua istilah ini.

## 6. Supervised parametric learning

Mesin parametrik punya **jumlah tombol (knobs) yang tetap**. Input diproses
berdasarkan posisi tombol, lalu menghasilkan prediksi. Belajar berarti
memutar tombol ke posisi lain. Seluruh pengetahuan model tersimpan dalam
posisi tombol.

Contoh buku: memprediksi peluang Red Sox memenangkan World Series.

Tiga langkahnya:
1. **Predict.** Data (misalnya statistik tim) diproses dan menghasilkan prediksi
   (misalnya 98%).
2. **Compare.** Prediksi dibandingkan dengan kenyataan. Jika tim kalah dan
   prediksinya 98%, maka selisihnya besar.
3. **Learn.** Tombol diputar berdasarkan seberapa besar kesalahannya dan data
   masukan yang dipakai. Tombol untuk input yang berguna dinaikkan; tombol
   untuk input yang tidak berguna diturunkan.

Buku menyebut proses ini sebagai *search*: mencari konfigurasi tombol yang
tepat dengan mencoba, menyesuaikan, dan mengulang.

## 7. Unsupervised parametric learning

Prinsipnya sama, tetapi tujuannya mengelompokkan data. Biasanya ada beberapa
tombol untuk setiap kelompok, dan tiap tombol menunjukkan seberapa cocok data
dengan kelompok tersebut. Hasil akhirnya adalah peluang keanggotaan tiap
kelompok (misalnya 94%, 1%, 5%).

## 8. Nonparametric learning

Nonparametrik belajar dengan **menghitung**. Jumlah parameter bertambah sesuai
jumlah hal yang dihitung dalam data.

Contoh buku: lampu lalu lintas. Model menghitung berapa kali mobil "jalan"
untuk setiap warna lampu. Setelah beberapa contoh, model dapat memprediksi
bahwa lampu tengah selalu (100%) membuat mobil jalan, sedangkan lampu kanan
hanya kadang-kadang (50%).

Buku menegaskan bahwa **deep learning adalah kelas model parametrik**.
Nonparametrik tidak dibahas lebih lanjut di buku.

## Key Ideas — Chapter 2

- Machine learning mengajarkan mesin meniru pola dari data.
- Supervised learning mengubah satu dataset menjadi dataset lain yang sudah diketahui jawabannya.
- Unsupervised learning mengelompokkan data tanpa jawaban yang sudah diketahui.
- Parametrik: jumlah parameter tetap, belajar dengan trial and error.
- Nonparametrik: jumlah parameter mengikuti data, belajar dengan menghitung.
- Deep learning adalah model parametrik yang belajar dengan siklus predict, compare, learn.

## Ilustrasi 1: Nonparametrik — Menghitung (bukan kode buku)

Contoh ini mengikuti deskripsi lampu lalu lintas di buku: lampu tengah
selalu membuat mobil jalan, lampu kanan hanya kadang-kadang. Angka pada
contoh di bawah dibuat untuk mencocokkan deskripsi tersebut.

```python
from collections import defaultdict

# Data: (lampu, jalan?) dengan 1 = mobil jalan, 0 = mobil berhenti
data = [('tengah', 1), ('tengah', 1), ('kanan', 1), ('kanan', 0)]

# Model nonparametrik: satu hitungan per lampu yang ditemukan dalam data
hitungan = defaultdict(lambda: {'jalan': 0, 'total': 0})
for lampu, jalan in data:
    hitungan[lampu]['jalan'] += jalan
    hitungan[lampu]['total'] += 1

for lampu, h in hitungan.items():
    peluang = h['jalan'] / h['total']
    print(f'lampu {lampu:7s}: peluang mobil jalan = {peluang:.0%}')

print('\nJumlah parameter (hitungan) =', len(hitungan), '-> mengikuti jumlah lampu dalam data')
```

## Ilustrasi 2: Parametrik — Memutar Tombol (bukan kode buku)

Contoh ini menunjukkan siklus predict, compare, learn dengan **satu tombol**.
Tombol di sini adalah angka `tombol`, yang diputar sedikit setiap langkah
untuk mengurangi kesalahan. Buku membahas ide ini dengan contoh tim baseball.
Pada chapter berikutnya, tombol ini disebut *weight*.

```python
tombol = 0.0          # posisi awal tombol
input_data = 2.0      # data masukan
kenyataan = 1.0       # hasil yang benar
laju = 0.1            # seberapa besar tombol diputar tiap langkah

for langkah in range(6):
    prediksi = input_data * tombol          # 1. Predict
    kesalahan = prediksi - kenyataan        # 2. Compare
    tombol -= laju * kesalahan * input_data # 3. Learn: putar tombol
    print(f'langkah {langkah}: prediksi = {prediksi:.3f}, kesalahan = {kesalahan:+.3f}')

print('posisi tombol akhir =', round(tombol, 3))
```
