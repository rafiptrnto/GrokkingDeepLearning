### Deep learning, machine learning, dan AI

Deep learning adalah **subset machine learning**, dan machine learning
adalah bagian dari bidang yang lebih luas (kecerdasan buatan). Deep
learning terutama memakai **artificial neural network**, yaitu kelas
algoritma yang terinspirasi secara longgar oleh otak manusia, dan dipakai
di industri untuk computer vision (gambar), NLP (teks), dan pengenalan
suara (audio).

### Machine learning

Machine learning adalah bidang yang membuat mesin belajar melakukan tugas
tanpa diprogram secara eksplisit: mesin mengamati pola dan menirunya
(analogi buku: *monkey see, monkey do*).

### Supervised vs unsupervised

- **Supervised learning**: meniru pola langsung antara dua dataset. Ia
  mengubah satu dataset menjadi dataset lain (misal harga saham Senin
  untuk memprediksi harga Selasa; piksel gambar untuk mendeteksi kucing).
  Ini "roti dan mentega" AI terapan.
- **Unsupervised learning**: mengelompokkan data. Tidak ada jawaban benar
  yang ditiru; algoritma mengubah data menjadi label cluster ("temukan
  pola pada data ini"). Semua bentuk unsupervised learning dapat dipandang
  sebagai bentuk clustering.

### Parametrik vs nonparametrik

- **Parametrik**: jumlah parameter ("tombol") **tetap**; belajar dengan
  trial and error, yaitu memutar tombol. Deep learning termasuk kelas ini.
- **Nonparametrik**: jumlah parameter **bergantung pada data**; umumnya
  berbasis penghitungan (counting).

Jadi ada empat kombinasi: supervised parametrik, unsupervised parametrik,
supervised nonparametrik, unsupervised nonparametrik.

### Tiga langkah supervised parametric learning

1. **Predict** — data masuk ke mesin dan keluar prediksi.
2. **Compare** — bandingkan prediksi dengan pola sebenarnya (kebenaran).
3. **Learn** — sesuaikan tombol berdasarkan seberapa besar kesalahan dan
   data masukan, agar prediksi berikutnya lebih akurat.

## Ilustrasi Kode

### 1. Model nonparametrik: menghitung (counting)

Contoh lampu lalu lintas pada buku: model hanya menghitung seberapa sering
mobil "jalan" untuk tiap warna lampu. Jumlah parameter (hitungan) mengikuti
jumlah kategori pada data.

```python
from collections import defaultdict

# (lampu, jalan?)  1 = mobil jalan, 0 = berhenti
data = [('kiri', 0), ('tengah', 1), ('kanan', 0), ('tengah', 1), ('kanan', 1), ('kiri', 0)]

jumlah = defaultdict(lambda: [0, 0])      # [jalan, total] per lampu
for lampu, jalan in data:
    jumlah[lampu][0] += jalan
    jumlah[lampu][1] += 1

for lampu, (jalan, total) in jumlah.items():
    print(f'lampu {lampu:7s}: peluang jalan = {jalan/total:.2f}  (parameter = hitungan {jalan}/{total})')
```

### 2. Model parametrik: memutar satu "tombol"

Satu parameter (weight) dengan trial and error: prediksi, bandingkan,
lalu putar tombol sedikit ke arah yang mengurangi kesalahan.

```python
# Belajar: output = 2 * input, lewat satu tombol (weight)
weight = 0.0
x, target = 3.0, 6.0

for langkah in range(20):
    pred = x * weight                 # 1. predict
    error = (pred - target) ** 2      # 2. compare
    turunan = 2 * (pred - target) * x
    weight -= 0.01 * turunan          # 3. learn
    if langkah % 4 == 0:
        print(f'langkah {langkah:2d}: weight={weight:.4f} pred={pred:.4f} error={error:.4f}')
print('weight akhir:', round(weight, 4))
```

## Kesimpulan


Machine learning dibagi dalam dua cara: berdasarkan jenis pola yang

parametrik yang memakai neural network untuk prediksi supervised maupun
unsupervised. Siklus **predict, compare, learn** pada chapter ini adalah
kerangka untuk seluruh chapter berikutnya.
