### Ringkasan konsep

- **Korelasi sebagai ringkasan (correlation summarization).** Neural
  network mencari korelasi antara input dan output; secara umum, jaringan
  ini adalah mesin yang **mencari dan menciptakan korelasi**. Gambaran
  tersederhana: input layer, hidden layer, dan output layer sebagai
  lapisan "node" yang dihubungkan weights.
- **Penyederhanaan visual.** Detail pada gambar jaringan (nilai tiap
  weight, tiap neuron) dapat dihilangkan. Yang perlu diingat hanyalah
  **layer sebagai vektor dan weights sebagai matriks** yang
  menghubungkannya (kotak = vektor, garis = matriks).
- **Notasi dengan huruf.** Layer dilambangkan huruf (`l0, l1, l2`) dan
  weights dengan `w`, sehingga jaringan dapat ditulis di atas kertas.
- **Menghubungkan variabel.** Perubahan pada satu variabel akan
  memengaruhi variabel lain yang terhubung (konsep "weight sebagai
  sensitivitas": jika input naik, seberapa naik output?).
- **Semua cocok bersama.** Satu jaringan dapat ditulis ringkas, misalnya
  `l2 = relu(l0.dot(w1)).dot(w2)`, dan bentuk dimensi (*shape*) tiap
  matriks harus saling cocok.

## Ilustrasi Kode

Jaringan tiga layer ditulis ringkas dengan notasi huruf: `l0` (input),
`l1` (hidden), `l2` (output), dan weights `w01`, `w12`. Perhatikan
bentuk (shape) tiap objek: layer = vektor, weights = matriks.

```python
import numpy as np
np.random.seed(1)

def relu(x):
    return (x > 0) * x

l0 = np.array([[1, 0, 1]])            # input: vektor 1x3 (satu data)
w01 = np.random.randn(3, 4)           # matriks 3x4 (3 input -> 4 hidden)
w12 = np.random.randn(4, 1)           # matriks 4x1 (4 hidden -> 1 output)

l1 = relu(l0.dot(w01))                # hidden layer: vektor 1x4
l2 = l1.dot(w12)                      # output: 1x1

print('l0  :', l0.shape)
print('w01 :', w01.shape)
print('l1  :', l1.shape)
print('w12 :', w12.shape)
print('l2  :', l2.shape, '->', l2)
```
