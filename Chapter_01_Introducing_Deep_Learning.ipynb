### Mengapa mempelajari deep learning

- **Alat untuk otomatisasi kecerdasan secara bertahap.** Deep learning
  menggunakan algoritma yang terinspirasi otak (neural network) dan
  mencapai rekor baru pada computer vision, pengenalan suara, dan
  terjemahan mesin.
- **Berpotensi mengotomatisasi pekerjaan terampil.** Contoh yang
  disebut buku: operator call-center, sopir taksi, dan analis bisnis
  tingkat dasar.
- **Menyenangkan dan kreatif.** Mencoba mensimulasikan kecerdasan dan
  kreativitas mengajarkan banyak hal tentang manusia, misalnya melatih
  mesin melukis seperti Monet.
